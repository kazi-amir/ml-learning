```python
numeric_cols = train_input.select_dtypes(include=np.number).columns.tolist()
numeric_cols
```

```python
categoric_cols = train_input.select_dtypes('str').columns.tolist()
categoric_cols
```

```python
from sklearn.impute import SimpleImputer
```

```python
imputer = SimpleImputer(strategy='mean')
```

```python
imputer.fit(data[numeric_cols])
```

```python
train_input[numeric_cols] = imputer.transform(train_input[numeric_cols])
val_input[numeric_cols] = imputer.transform(val_input[numeric_cols])
test_input[numeric_cols] = imputer.transform(test_input[numeric_cols])
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()
```

```python
scaler.fit(data[numeric_cols])
```

```python
train_input[numeric_cols] = scaler.transform(train_input[numeric_cols])
val_input[numeric_cols] = scaler.transform(val_input[numeric_cols])
test_input[numeric_cols] = scaler.transform(test_input[numeric_cols])
```

```python
train_input[categoric_cols].nunique()
```

```python
cate_no_na = data[categoric_cols].fillna('Unknown')
cate_no_na
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
```

```python
encoder.fit(cate_no_na)
```

```python
encoded_cols = list(encoder.get_feature_names_out(categoric_cols))
print(encoded_cols)
```

```python
# 1. Transform and wrap the encoded results into new DataFrames
# Ensure the index matches the original DataFrame so they align perfectly
train_encoded = pd.DataFrame(
    encoder.transform(train_input[categoric_cols]), 
    columns=encoded_cols, 
    index=train_input.index
).fillna('Unknown')

val_encoded = pd.DataFrame(
    encoder.transform(val_input[categoric_cols]), 
    columns=encoded_cols, 
    index=val_input.index
).fillna('Unknown')

test_encoded = pd.DataFrame(
    encoder.transform(test_input[categoric_cols]), 
    columns=encoded_cols, 
    index=test_input.index
).fillna('Unknown')

# 2. Combine them cleanly using pd.concat
train_input = pd.concat([train_input, train_encoded], axis=1)
val_input = pd.concat([val_input, val_encoded], axis=1)
test_input = pd.concat([test_input, test_encoded], axis=1)
```

```python
 train_input
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression(solver='liblinear')
```

```python
model.fit(train_input[numeric_cols + encoded_cols], train_target)
```

```python
X_train = train_input[numeric_cols + encoded_cols]
X_val = val_input[numeric_cols + encoded_cols]
X_test = test_input[numeric_cols + encoded_cols]
```

```python
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
```

```python
def predict_and_plot(inputs, targets, name=''):
    preds = model.predict(inputs)
    
    accuracy = accuracy_score(targets, preds)
    print("Accuracy: {:.2f}%".format(accuracy * 100))
    
    cf = confusion_matrix(targets, preds, normalize='true')
    plt.figure()
    sns.heatmap(cf, annot=True)
    plt.xlabel('Prediction')
    plt.ylabel('Target')
    plt.title('{} Confusion Matrix'.format(name));
    
    return preds
```

```python
train_predict = predict_and_plot(X_train, train_target, 'Training Data')
```

```python
val_predict = predict_and_plot(X_val, val_target, 'Validation Data')
```

```python
test_predict = predict_and_plot(X_test, test_target, 'Test Data')
```

```python
def predict_input(single_input):
    # 1. Create base DataFrame
    input_df = pd.DataFrame([single_input])
    
    # 2. Process numeric columns in-place (low number of columns, no fragmentation risk)
    input_df[numeric_cols] = imputer.transform(input_df[numeric_cols])
    input_df[numeric_cols] = scaler.transform(input_df[numeric_cols])
    
    # 3. Create a clean standalone DataFrame for the one-hot encoded columns
    encoded_df = pd.DataFrame(
        encoder.transform(input_df[categoric_cols]),
        columns=encoded_cols,
        index=input_df.index
    )
    
    # 4. Concatenate them horizontally to build X_input safely
    X_input = pd.concat([input_df[numeric_cols], encoded_df], axis=1)
    
    # 5. Predict
    pred = model.predict(X_input)[0]
    prob = model.predict_proba(X_input)[0][list(model.classes_).index(pred)]
    
    return pred, prob
```
