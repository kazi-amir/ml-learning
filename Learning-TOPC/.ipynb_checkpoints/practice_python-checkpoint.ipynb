# Basics

```python
def find_min(List):
    minValue = List[0]
    minIndex = 0
    idx = 1;
    n = len(List)
    while(idx < n):
        if(List[idx] < minValue):
            minValue = List[idx]
            minIndex = idx
        idx+=1

    return minValue, minIndex
```

```python
def selection_sort():
    sorted_list = []
    while(len(L) > 0):
        minV, minI = find_min(L)
        sorted_list.append(minV)
        L.pop(minI)
    return sorted_list
```

```python
L = [2, 5, 3, 1, 9, 0]
n = len(L)
sorted_list = selection_sort()
print(sorted_list)
```

## Simple Imputer

```python
from sklearn.impute import SimpleImputer
```

```python
df = pd.read_csv('topc_dataset.csv')
```

```python
df.head
```

```python
df.drop(['Lab teacher name','Tshirt size','Transportation service','Gender','Payment method','Payment phone','Status','Contest id'
], axis=1,inplace=True)
df.head()
```

```python
df.rename(columns={'Timestamp':'Date','Student id':'Student ID', 'Payment transaction id':'TRXID', 'User id':'User ID'}, inplace=True)
df.head()
```

```python
for column in df.columns:
    df[column] = df[column].astype(str)

df.head()
```

```python
df['Date'] = df['Date'].str[:10]
df.head()
```

```python
df['Date'] = pd.to_datetime(df['Date'])
df.head()
```

```python
df.describe()
```

```python
df.columns
```

```python
df.info()
```

```python
df.fillna('NA', inplace=True)
```

```python
df.head()
```

```python
df2 = df.groupby(['Department', 'Section'])[['Pickup point']].count()
```
