```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
```

```python
data = pd.read_csv('house_prices_grad_desc.csv')
```

```python
data.head()
```

```python
ndata = data[['LotArea', 'SalePrice']].copy()
ndata
```

```python
ndata['LotArea'] = ndata['LotArea'].fillna(ndata['LotArea'].mean(), inplace = False)
```

```python
ndata['SalePrice'] = ndata['SalePrice'].fillna(ndata['SalePrice'].mean(), inplace = False)
```

```python
lotArea = np.array(ndata['LotArea'])
lotArea
```

```python
salePrice = np.array(ndata['SalePrice'])
salePrice
```

## Standardising the data (Z-score normalization)

```python
lotArea = (lotArea - np.mean(lotArea)) / np.std(lotArea)
```

## y = b + wx

$y = h_w(x)$ - models predicted output y

$w_0 = b$

$w_1 = w$

$h_w(x) = w_0 + w_1 x$

```python
def predict(x , w, b):
    return b + (w * x)
```

## Cost Function for one data point: $Cost = \{h_w(x) - Y\}^2$
## Cost Function for the whole dataset: $j(w_0, w_1) = \frac{1}{2m} \sum_{i=1}^m \{h_w(x_i) - y_i\}^2$

```python
def cost_function(lotArea, salePrice, w, b):
    m = len(lotArea)
    error = 0.0
    for i in range(m):
        error += (predict(lotArea[i], w, b) - salePrice[i])**2
    return error / (2 * m)
```

## Weight Update

$w\_derivative = \frac{1}{m} \sum_{i = 1}^m \{h_w(x_i) - y_i\} \cdot x_i$

$b\_derivative = \frac{1}{m} \sum_{i = 1}^m \{h_w(x_i) - y_i\}$

$param_j = param_j - \alpha \frac{d}{dparam_j}j(w, b);\ \alpha = learning\ rate$

```python
def update_weight(lotArea, salePrice, w, b, learning_rate):
    w_deriv = 0
    b_deriv = 0
    m = len(lotArea)
    for i in range(m):
        w_deriv += (predict(lotArea[i], w, b) - salePrice[i]) * lotArea[i]
        b_deriv += (predict(lotArea[i], w, b) - salePrice[i])

    # Update
    w -= learning_rate * (w_deriv / m)
    b -= learning_rate * (b_deriv / m)
    
    return w, b
```

## Model training phase (Loop)

```python
w = 5.0
b = 5.0
learning_rate = 0.01
epochs = 9000
tolerance = 1.0

prev_cost = cost_function(lotArea, salePrice, w, b)

for epoch in range(epochs):
    w, b = update_weight(lotArea, salePrice, w, b, learning_rate)

    current_cost = cost_function(lotArea, salePrice, w, b)
    
    if epoch % 50 == 0 or epoch == 1:
        print(f"Epoch {epoch}: Cost = {current_cost:.5f}")
        
    cost_decrease = prev_cost - current_cost
    
    if tolerance > cost_decrease:
        print(f"\n[Convergence Reached] Stopped early at Epoch {epoch}!")
        print(f"The cost only decreased by {cost_decrease:.5f} (which is less than our tolerance of {tolerance})")
        print(f"Final Cost: {current_cost:,.5f}")
        break

    prev_cost = current_cost
    
print(f"Weight: {w:.5f}, Bias: {b:.5f}")
```

```python
lot_mean = np.mean(ndata['LotArea'])
lot_std = np.std(ndata['LotArea'])

# 2. Scale the test data
test_data = 5000
test_data_scaled = (test_data - lot_mean) / lot_std

# 3. Make the manual prediction
prediction = predict(test_data_scaled, w, b)


print(f"Manual Predicted Price for 5,000 sqft lot: ${prediction:,.5f}")
```

## Built-in Gradient Descent Linear Regression Model

```python
from sklearn.linear_model import SGDRegressor
from sklearn.preprocessing import StandardScaler
```

```python
X = np.array(ndata['LotArea']).reshape(-1, 1)
y = np.array(ndata['SalePrice'])
```

```python
scaler_X = StandardScaler()
X_scaled = scaler_X.fit_transform(X)
```

```python
sgd_model = SGDRegressor(max_iter=1000, tol=1e-3, eta0=0.01, random_state=42)
sgd_model.fit(X_scaled, y)
```

```python
w_scikit = sgd_model.coef_[0]
b_scikit = sgd_model.intercept_[0]
```

```python
print(f"{w_scikit:.5f} {b_scikit:.5f}")
```

```python
# 1. New house area
new_house = np.array([[5000]]) 

# 2. Scale the input using the same scaler from training
new_house_scaled = scaler_X.transform(new_house)

# 3. Predict the scaled price
predicted_price = sgd_model.predict(new_house_scaled)

print(f"Predicted price for 5,000 sqft lot: ${predicted_price[0]:.5f}")
```
