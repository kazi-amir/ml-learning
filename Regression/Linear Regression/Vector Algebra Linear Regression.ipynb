# Vector Algebra Linear Regression - Manual Inplementation

```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
```

```python
data = pd.read_csv('house_prices_grad_desc.csv')
data.head()
```

```python
features = ['LotFrontage', 'LotArea', 'TotalBsmtSF', 'GarageArea']
target = 'SalePrice'
```

```python
for feature in features:
    data[feature] = data[feature].fillna(data[feature].mean())
```

```python
X_raw = data[features].values
X_raw
```

```python
Y = data[target].values
Y
```

```python
# Number of data point
m = len(Y)
```

```python
ones_column = np.ones((m, 1))
```

```python
X_matrix = np.append(ones_column, X_raw, axis = 1)
X_matrix
```

## Matrix Calculations
## $X^t$

```python
X_t = X_matrix.T
X_t
```

## $X^tX$

```python
X_t_X = np.dot(X_t, X_matrix)
X_t_X
```

## $(X^tX)^{-1}$

```python
X_t_X_inv = np.linalg.inv(X_t_X)
X_t_X_inv
```

## $X^tY$

```python
X_t_Y = np.dot(X_t, Y)
X_t_Y
```

## w = $(X^tX)^{-1}X^tY$

```python
w = np.dot(X_t_X_inv, X_t_Y)
w
```

```python
print("\nCalculation Complete!")
print("-" * 50)
print(f"Exact Bias (w_0)        : {w[0]:.2f}")
for j in range(len(features)):
    print(f"Exact Weight {features[j]} : {w[j+1]:.5f}")
```

# Prediction:
$Y = w^t x$

```python
test_data = np.array([80, 9500, 1500, 700])

test_data_with_bias = np.insert(test_data, 0, 1)

w_t = w.T

exact_prediction = np.dot(w_t, test_data_with_bias)

print(f"\nNormal Equation Prediction: ${exact_prediction:,.2f}")
```
